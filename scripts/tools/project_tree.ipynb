```python
import os


# 除外するフォルダ
IGNORE_DIRS = {
    "__pycache__",
    ".git",
    ".idea",
    ".vscode",
    ".pytest_cache",
    ".mypy_cache",
}

# プロジェクトルートからの相対パスで除外
IGNORE_PATHS = {
    "output/img/machines",
    "output/machines",
    "output/makers",    
    "output/gigaslot/img",   
    "static/img/machines"
}


def print_tree(root_path):

    root_path = os.path.abspath(root_path)

    print(os.path.basename(root_path) + "/")

    def _walk(current_path, prefix=""):

        rel_path = os.path.relpath(
            current_path,
            root_path,
        ).replace("\\", "/")

        # 指定パスは中身ごと除外
        if rel_path in IGNORE_PATHS:
            return

        entries = []

        for name in os.listdir(current_path):

            if name.startswith("."):
                continue

            full_path = os.path.join(
                current_path,
                name,
            )

            # __pycache__等は除外
            if (
                os.path.isdir(full_path)
                and name in IGNORE_DIRS
            ):
                continue

            entries.append(name)

        # フォルダ→ファイル順
        entries.sort(
            key=lambda x: (
                not os.path.isdir(
                    os.path.join(current_path, x)
                ),
                x.lower(),
            )
        )

        for i, name in enumerate(entries):

            full_path = os.path.join(
                current_path,
                name,
            )

            is_last = (
                i == len(entries) - 1
            )

            branch = (
                "└── "
                if is_last
                else "├── "
            )

            print(
                prefix + branch + name
            )

            if os.path.isdir(full_path):

                extension = (
                    "    "
                    if is_last
                    else "│   "
                )

                _walk(
                    full_path,
                    prefix + extension,
                )


    _walk(root_path)


# ===========================
# 実行
# ===========================

try:
    base_dir = os.path.dirname(
        os.path.abspath(__file__)
    )
except NameError:
    base_dir = os.getcwd()

project_root = os.path.abspath(
    os.path.join(
        base_dir,
        "..",
        "..",
    )
)

print_tree(project_root)
```
