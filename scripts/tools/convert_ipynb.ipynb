```python
import subprocess
import sys
from pathlib import Path


# ==================================================
# パス設定
# ==================================================

try:
    BASE_DIR = Path(__file__).resolve().parent
except NameError:
    BASE_DIR = Path.cwd()

print(f"[INFO] 対象フォルダ: {BASE_DIR}")


# ==================================================
# 設定
# ==================================================

SELF_NOTEBOOK = "convert_ipynb.ipynb"


# ==================================================
# 変換
# ==================================================

for notebook_path in sorted(
    BASE_DIR.glob("*.ipynb")
):

    # 自分自身は除外
    if notebook_path.name == SELF_NOTEBOOK:
        continue

    print(f"[INFO] 変換: {notebook_path.name}")

    subprocess.run(
        [
            sys.executable,
            "-m",
            "jupyter",
            "nbconvert",
            "--to",
            "script",
            str(notebook_path),
            "--output-dir",
            str(BASE_DIR),
        ],
        check=True,
    )

print("[INFO] 完了")
```
