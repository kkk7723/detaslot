```python
from ftplib import FTP_TLS
import os

# ===== 共通設定 =====
ftp_host = "sv16194.xserver.jp"
ftp_user = "sedoinfinity"
ftp_pass = "d958b40323"

today = "20250716"

# ===== ローカルパス =====
if os.name == 'nt':
    user_base = os.path.join(os.environ["USERPROFILE"], "myenv", "domz")
else:
    user_base = os.path.join(os.path.expanduser("~"), "myenv", "domz")

local_html_dir = user_base
local_webp_dir = os.path.join(user_base, "pic", today, "webp")

# ===== リモートパス =====
remote_html_dir = "sedoinfinity.xsrv.jp/public_html/domz/"
remote_webp_dir = f"sedoinfinity.xsrv.jp/public_html/domz/pic/{today}/webp/"

# ✅ FTPS接続
ftps = FTP_TLS(ftp_host)
ftps.login(ftp_user, ftp_pass)
ftps.prot_p()
print("[INFO] ✅ FTPS接続完了")


# ✅ HTMLアップロード（HTMLファイルのみ）
def upload_html(local_dir, remote_dir):
    print(f"[INFO] ✅ HTMLアップロード開始 {local_dir} → {remote_dir}")

    exclude_dirs = ["pic", "db", "templates", "scripts", ".ipynb_checkpoints"]

    ftps.cwd("/")
    for part in remote_dir.strip("/").split("/"):
        ftps.cwd(part)

    for root, dirs, files in os.walk(local_dir):
        # ✅ 除外ディレクトリ除外
        if any(skip in root for skip in exclude_dirs):
            continue

        rel_path = os.path.relpath(root, local_dir)

        # ✅ サブフォルダ作成
        if rel_path != ".":
            for part in rel_path.split(os.sep):
                try:
                    ftps.mkd(part)
                except Exception:
                    pass
                ftps.cwd(part)

        # ✅ HTMLファイルのみアップロード
        for file in files:
            if not file.endswith(".html"):
                continue
            local_file = os.path.join(root, file)
            with open(local_file, 'rb') as f:
                ftps.storbinary(f'STOR ' + file, f)
            print(f"[INFO] ✅ アップロード: {file}")

        # ✅ 上の階層へ戻る
        if rel_path != ".":
            for _ in rel_path.split(os.sep):
                ftps.cwd("..")

    ftps.cwd("/")
    print("[INFO] ✅ HTMLアップロード完了\n")


upload_html(local_html_dir, remote_html_dir)


# ✅ webpアップロード
def upload_webp(local_dir, remote_dir):
    print(f"[INFO] ✅ webpフォルダアップロード開始 {local_dir} → {remote_dir}")
    ftps.cwd("/")
    for part in remote_dir.strip("/").split("/"):
        try:
            ftps.cwd(part)
        except Exception:
            ftps.mkd(part)
            ftps.cwd(part)

    for filename in os.listdir(local_dir):
        path = os.path.join(local_dir, filename)
        if os.path.isfile(path):
            with open(path, 'rb') as f:
                ftps.storbinary(f'STOR ' + filename, f)
            print(f"[INFO] ✅ アップロード: {filename}")

    ftps.cwd("/")
    print("[INFO] ✅ webpアップロード完了\n")


upload_webp(local_webp_dir, remote_webp_dir)

ftps.quit()
print("[INFO] ✅ ✅ アップロード＆切断 完了")
```

```python
from ftplib import FTP_TLS, error_perm
import os
import time

# ===== 共通設定 =====
ftp_host = "sv16194.xserver.jp"
ftp_user = "sedoinfinity"
ftp_pass = "d958b40323"

today = "20250716"

# ===== ローカルパス =====
if os.name == 'nt':
    user_base = os.path.join(os.environ["USERPROFILE"], "myenv", "domz")
else:
    user_base = os.path.join(os.path.expanduser("~"), "myenv", "domz")

local_html_dir = os.path.normpath(user_base)
local_webp_dir = os.path.normpath(os.path.join(user_base, "pic", today, "webp"))

# ===== リモートパス =====
remote_html_dir = "/sedoinfinity.xsrv.jp/public_html/domz"
remote_webp_dir = f"/sedoinfinity.xsrv.jp/public_html/domz/pic/{today}/webp"


def connect_ftps():
    ftps = FTP_TLS(ftp_host)
    ftps.login(ftp_user, ftp_pass)
    ftps.prot_p()
    ftps.set_pasv(True)
    print("[INFO] ✅ FTPS接続成功")
    return ftps


def make_remote_dirs(ftps, remote_dir):
    ftps.cwd("/")
    for part in remote_dir.strip("/").split("/"):
        try:
            ftps.cwd(part)
        except error_perm:
            ftps.mkd(part)
            ftps.cwd(part)


# ✅ HTMLアップロード（HTMLファイルのみ）
def upload_html(local_dir, remote_dir, max_retries=3):
    print(f"[INFO] ▶️ HTMLアップロード開始 {local_dir} → {remote_dir}")
    exclude_dirs = ["pic", "db", "templates", "scripts", ".ipynb_checkpoints"]

    for attempt in range(1, max_retries + 1):
        try:
            with connect_ftps() as ftps:
                make_remote_dirs(ftps, remote_dir)

                for root, dirs, files in os.walk(local_dir):
                    # 除外ディレクトリ除外
                    if any(skip in root for skip in exclude_dirs):
                        continue

                    rel_path = os.path.relpath(root, local_dir).replace(os.sep, '/')
                    target_dir = remote_dir if rel_path == '.' else f"{remote_dir}/{rel_path}"

                    make_remote_dirs(ftps, target_dir)

                    # HTMLファイルアップロード
                    for file in files:
                        if file.endswith(".html"):
                            local_file = os.path.join(root, file)
                            with open(local_file, 'rb') as f:
                                ftps.storbinary(f'STOR {file}', f)
                            print(f"[INFO] ✅ アップロード: {file}")

                print("[INFO] ✅ HTMLアップロード完了\n")
            break
        except Exception as e:
            print(f"[ERROR] ❌ HTMLアップロード失敗({attempt}/{max_retries}): {e}")
            if attempt < max_retries:
                time.sleep(3)
            else:
                print("[FATAL] 🚫 HTMLアップロード失敗。スキップ")


# ✅ webpフォルダアップロード
def upload_webp(local_dir, remote_dir, max_retries=3):
    print(f"[INFO] ▶️ webpフォルダアップロード開始 {local_dir} → {remote_dir}")
    for attempt in range(1, max_retries + 1):
        try:
            with connect_ftps() as ftps:
                make_remote_dirs(ftps, remote_dir)

                for filename in os.listdir(local_dir):
                    local_file = os.path.join(local_dir, filename)
                    if os.path.isfile(local_file):
                        with open(local_file, 'rb') as f:
                            ftps.storbinary(f'STOR {filename}', f)
                        print(f"[INFO] ✅ アップロード: {filename}")

                print("[INFO] ✅ webpアップロード完了\n")
            break
        except Exception as e:
            print(f"[ERROR] ❌ webpアップロード失敗({attempt}/{max_retries}): {e}")
            if attempt < max_retries:
                time.sleep(3)
            else:
                print("[FATAL] 🚫 webpアップロード失敗。スキップ")


if __name__ == '__main__':
    upload_html(local_html_dir, remote_html_dir)
    upload_webp(local_webp_dir, remote_webp_dir)
    print("[INFO] ✅ ✅ 全アップロード完了")
```
